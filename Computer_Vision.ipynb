```python
from google.colab import files

uploaded = files.upload()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]

image = cv2.imread(filename)
```

```python
print("Image uploaded successfully:", filename)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
colors = ['b', 'g', 'r']

for i, color in enumerate(colors):
    histogram = cv2.calcHist([image], [i], None, [256], [0, 256])
    plt.plot(histogram, color=color)

plt.title("Color Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.show()
```

```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

slow_video = cv2.VideoWriter(
    'slow_motion.mp4',
    fourcc,
    fps / 2,
    (width, height)
)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    slow_video.write(frame)

cap.release()
slow_video.release()

print("Slow-motion video created")
```

```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

fast_video = cv2.VideoWriter(
    'fast_motion.mp4',
    fourcc,
    fps,
    (width, height)
)

frame_count = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Write every second frame
    if frame_count % 2 == 0:
        fast_video.write(frame)

    frame_count += 1

cap.release()
fast_video.release()

print("Fast-motion video created")
```
