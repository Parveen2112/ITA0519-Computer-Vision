```python
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

fast_video = cv2.VideoWriter(
    'fast_motion.mp4',
    fourcc,
    fps,
    (width, height)
)

frame_count = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Write every second frame
    if frame_count % 2 == 0:
        fast_video.write(frame)

    frame_count += 1

cap.release()
fast_video.release()

print("Fast-motion video created")
```

```python
height, width = image.shape[:2]
print("Width:", width)
print("Height:", height)
```

```python
source_points = np.float32([
    [50, 50],
    [200, 50],
    [50, 200]
])

destination_points = np.float32([
    [20, 80],
    [220, 50],
    [80, 220]
])
```

```python
matrix = cv2.getAffineTransform(source_points, destination_points)
print("Affine Matrix:")
print(matrix)
```

```python
affine_image = cv2.warpAffine(
    image,
    matrix,
    (width, height)
)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(affine_image, cv2.COLOR_BGR2RGB))
plt.title("Affine Transformed Image")
plt.axis("off")

plt.show()
```

```python
height, width = image.shape[:2]
print("Width:", width)
print("Height:", height)

```

```python
source_points = np.float32([
    [50, 50],
    [width - 50, 50],
    [50, height - 50],
    [width - 50, height - 50]
])

destination_points = np.float32([
    [0, 0],
    [width, 0],
    [100, height],
    [width - 100, height]
])
```

```python
matrix = cv2.getPerspectiveTransform(
    source_points,
    destination_points
)
print("Perspective Matrix:")
print(matrix)
```

```python
perspective_image = cv2.warpPerspective(
    image,
    matrix,
    (width, height)
)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(perspective_image, cv2.COLOR_BGR2RGB))
plt.title("Perspective Transformed Image")
plt.axis("off")

plt.show()

```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]
image = cv2.imread(filename)
print("Image uploaded successfully:", filename)

```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

gray = np.float32(gray)

corners = cv2.cornerHarris(
    gray,
    blockSize=2,
    ksize=3,
    k=0.04
)
```

```python
result = image.copy()
result[corners > 0.01 * corners.max()] = [0, 0, 255]

```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Harris Corner Detection")
plt.axis("off")
plt.show()
```

**LAB16**

```python
from google.colab import files
uploaded = files.upload()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]
image = cv2.imread(filename)
print("Image uploaded successfully:", filename)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

sobel_x = cv2.Sobel(
    gray,
    cv2.CV_64F,
    1,
    0,
    ksize=3
)


sobel_y = cv2.Sobel(
    gray,
    cv2.CV_64F,
    0,
    1,
    ksize=3
)
```

```python
sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

sobel_combined = cv2.addWeighted(
    sobel_x, 0.5,
    sobel_y, 0.5,
    0
)
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.subplot(2, 2, 2)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X - Vertical Edges")
plt.axis("off")
plt.subplot(2, 2, 3)
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y - Horizontal Edges")
plt.axis("off")
plt.subplot(2, 2, 4)
plt.imshow(sobel_combined, cmap="gray")
plt.title("Combined Sobel Edge Detection")
plt.axis("off")
plt.show()

```

```python
from google.colab import files
uploaded = files.upload()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions
```

```python
filename = list(uploaded.keys())[0]
image = cv2.imread(filename)
print("Image uploaded successfully:", filename)
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Input Image")
plt.axis("off")
plt.show()
```

```python
model = MobileNetV2(weights="imagenet")
img = cv2.resize(image, (224, 224))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = np.expand_dims(img, axis=0)
img = preprocess_input(img)
predictions = model.predict(img)
results = decode_predictions(predictions, top=5)[0]
```

```python
for _, label, confidence in results:
    print(label, ":", round(confidence * 100, 2), "%")
best_label = results[0][1]
confidence = results[0][2] * 100
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title(f"Detected: {best_label} ({confidence:.2f}%)")
plt.axis("off")
plt.show()

```
